# Retrieving steam_sales data:

```python
import requests
import pandas as pd

url = "https://api.gamalytic.com/steam-games/list"

all_games = []
page = 0

while True:
    params = {
        "limit": 1000,
        "page": page
    }

    response = requests.get(url, params=params)
    response.raise_for_status()

    data = response.json()
    games = data['result']

    if len(games) == 0:
        break

    all_games.extend(games)

    print(f"Page {page}: {len(games)} games")

    page += 1

df_sales = pd.DataFrame(all_games)
```
