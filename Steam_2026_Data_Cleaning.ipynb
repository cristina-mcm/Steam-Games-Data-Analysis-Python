### As we can see, the genres, categories and tags columns have the same information stored in different formats. We want to change the values that is a list of dictionaries.

### Function to turn the text inside the column to a consistent Python list:

```python
def clean (value):
    if pd.isna(value):
        return []

    try:
        parsed = ast.literal_eval(value)

        # Format: ["Action", "Adventure"]
        if isinstance(parsed, list) and all(isinstance(x, str) for x in parsed):
            return parsed

        # Format: [{"id": "1", "description": "Action"}, ...]
        if isinstance(parsed, list) and all(isinstance(x, dict) for x in parsed):
            return [x.get("description") for x in parsed if x.get("description")]

    except:
        return []

    return []
```

### Clean the genre column

```python
df2['genres_clean'] = df2['genres'].apply(clean)

df2['genres_clean'].explode().value_counts()
```

### The genres_clean column contains genres in different languages. We need to fix that. Later, we will do the same with the categories_clean column.

```python
genre_replacements = {
    'Aventure': 'Adventure',
    'Occasionnel': 'Casual',
    'Indépendant': 'Indie',
    'Stratégie': 'Strategy',
    'Aventura': 'Adventure',
    'Simuladores': 'Simulation',
    'インディー': 'Indie',
    '早期アクセス': 'Early Access',
    'Strategie': 'Strategy',
    'Gratuit': 'Free To Play'
}
```

```python
df2['genres_clean'] = df2['genres_clean'].apply(lambda genres: [genre_replacements.get(genre, genre) for genre in genres])
```

### After merging the two datasets using a left join, I created a separate DataFrame containing only the games that had matching records in both datasets. Games without a corresponding steamId from the Gamalytic dataset were excluded from df_matched.

```python
df_matched = df2[df2['steamId'].notna()].copy() #df2['steamId'].notna() checks which rows have a non-missing steamId
df_matched.shape
```

### Finally, I deleted the repeated columns and the ones that had the old useless information
